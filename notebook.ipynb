```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv('data/Housing.csv')
print(df)
```

```python
# Data Preprocessing
#df = pd.get_dummies(df, columns=['furnishingstatus'], drop_first=True)
yes_no = [col for col in df.columns if df[col].isin(['yes','no']).all()]
df[yes_no] = df[yes_no].replace({'yes': 1, "no": 0})
df = df.astype(float)
df.head(1)
```

```python
# Visualize the relationship between each feature and the target variable (price)
for col in df.columns:
    plt.scatter(df[col], df.price)
    plt.title(f"Scatter Plot between {col} and Prices")
    plt.xlabel(col)
    plt.ylabel("Price")
    plt.show()
```

```python
# Initializating X and y
X_train = df.drop(columns=['price']).to_numpy()
y_train = df.price.to_numpy()
m = X_train.shape[0]
print(X_train)
print(y_train)
```

```python
# Feature Scaling
X_mean = X_train.mean(axis=0)
X_std = X_train.std(axis=0)
X_train = (X_train - X_mean) / X_std
print(X_train)
```

```python
# Linear Regression Model
def linear_regression(w, b, X):
    return np.dot(X, w) + b
```

```python
# Cost function MSE
def mean_squared_error(y_true, y_pred):
    return 0.5 * np.mean((y_true - y_pred) ** 2)
```

```python
# Gradient Descent
def gradient_descent(w, b, X, y ,learning_rate):
    y_pred = linear_regression(w, b, X)
    error = y_pred - y
    dj_dw = np.dot(X.T, error) / len(y)
    dj_db = np.mean(error)

    w = w - (learning_rate*dj_dw)
    b = b - (learning_rate*dj_db)
    return w,b
```

```python
# Training the model
w = np.zeros(X_train.shape[1])
b = 0
learning_rate = 0.01
num_epochs = 5000
losses = []
for curr_epoch in range(num_epochs):
    w, b = gradient_descent(w, b, X_train, y_train, learning_rate)
    y_pred = linear_regression(w, b, X_train)
    loss = mean_squared_error(y_train, y_pred)
    losses.append(loss)
    if len(losses) > 1 and abs(losses[-2] - losses[-1]) < 1e-6:
        print("Convergence reached.")
        print(f"Epoch: {curr_epoch}, Loss: {loss}")
        break
```

```python
print(w, b)
print(losses)
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```

```python
def predict(X_new, w, b, X_mean, X_std):
    X_new = (X_new - X_mean) / X_std
    return linear_regression(w, b, X_new)
```

```python
house = np.array([5000, 4, 2, 3, 1, 0, 1, 1, 0, 1, 2, 0, 1])
predicted_price = predict(house, w, b, X_mean, X_std)
print(f"Predicted price for the house: {predicted_price}")
```

```python
y_pred_all = linear_regression(w, b, X_train)
mae = np.mean(np.abs(y_train - y_pred_all))
print(f"Mean Absolute Error on training data: {mae:,.0f}")
print(f"Mean price: {y_train.mean():,.0f}")
print(f"Min price: {y_train.min():,.0f}")
print(f"Max price: {y_train.max():,.0f}")
```

# Linear Regression with sklearn
- model.fit() -> training loop
- model.predict() -> the linear regression model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

model = LinearRegression()
model.fit(X_train, y_train)
print(model.coef_, model.intercept_)

y_pred = model.predict(X_train)
mae = mean_absolute_error(y_train, y_pred)
print(f"Mean Absolute Error on training data: {mae:,.0f}")
print(f"Mean price: {y_pred.mean():,.0f}")
print(f"Min price: {y_pred.min():,.0f}")
print(f"Max price: {y_pred.max():,.0f}")
```
